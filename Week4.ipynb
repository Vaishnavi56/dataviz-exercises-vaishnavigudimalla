```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

```python
# Task 1 — Scatter: life expectancy change over time (with movement lines)

# Choose one continent
continent = "Europe"

# Filter data for 2000 and 2007
df_scatter = df.loc[
    (df['continent'] == continent) &
    (df['year'].isin([2000, 2007]))
].sort_values(by=["country", "year"])

# Create scatter with lines connecting years
fig = px.line(
    df_scatter,
    x="gdpPercap",
    y="lifeExp",
    color="year",  # still only 2 colors
    line_group="country",  # connects 2000 → 2007
    hover_name="country",
    log_x=True,
    markers=True,
    title=f"{continent}: Movement in Life Expectancy vs Income (2000 → 2007)",
    labels={
        "gdpPercap": "GDP per Capita (log scale)",
        "lifeExp": "Life Expectancy",
        "year": "Year"
    },
    color_discrete_map={
        2000: "steelblue",
        2007: "darkorange"
    }
)

fig.update_traces(marker=dict(size=6))

fig.show()
```

```python
# Task 1 — Scatter: life expectancy change over time (2002 → 2007)

# Choose one continent
continent = "Europe"

# Filter data for 2002 and 2007
df_scatter = df.loc[
    (df['continent'] == continent) &
    (df['year'].isin([2002, 2007]))
].sort_values(by=["country", "year"])

# Create scatter with movement lines
fig = px.line(
    df_scatter,
    x="gdpPercap",
    y="lifeExp",
    color="year",  # 2 colors only
    line_group="country",  # connect points per country
    hover_name="country",
    log_x=True,
    markers=True,
    title=f"{continent}: Countries Improved in Income and Life Expectancy (2002 → 2007)",
    labels={
        "gdpPercap": "GDP per Capita (log scale)",
        "lifeExp": "Life Expectancy",
        "year": "Year"
    },
    color_discrete_map={
        2002: "steelblue",
        2007: "darkorange"
    }
)

fig.update_traces(marker=dict(size=6))

fig.show()
```

```python
# Task 2 — Bubble chart: tell a story

# Filter for 2007 data only
df_2007 = df.loc[df['year'] == 2007].copy()

# Choose one continent to highlight
highlight_continent = "Asia"

# Create a helper column for coloring
df_2007["highlight"] = df_2007["continent"].apply(
    lambda x: highlight_continent if x == highlight_continent else "Other"
)

# Bubble chart
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="highlight",
    hover_name="country",
    log_x=True,
    size_max=60,
    title=f"2007 Global Health vs Wealth — {highlight_continent} Shows Wide Development Range",
    labels={
        "gdpPercap": "GDP per Capita (log scale)",
        "lifeExp": "Life Expectancy",
        "highlight": "Continent"
    },
    color_discrete_map={
        highlight_continent: "crimson",
        "Other": "skyblue"
    }
)

# Add annotation
fig.add_annotation(
    x=1500,
    y=65,
    text=f"{highlight_continent}: Large variation — from low-income countries to rapidly developing economies.",
    showarrow=True,
    arrowhead=2
)

fig.show()
```
